#### 3 · Lanzar tus propias excepciones

Una función también puede **lanzar** una excepción con `raise`, cuando recibe algo con lo que no puede trabajar. La excepción **se propaga**: sale de la función y la atiende quien la llamó.

````python
def raiz(x):
    if x < 0:
        raise ValueError("número negativo")
    return x ** 0.5

try:
    r = raiz(-4)
except ValueError as e:
    r = None
    print("Error:", e)
print(r)
````

#### 4 · `else`, `finally` y el orden de los `except`

Un `try` completo puede tener cuatro partes:

````python
def dividir(a, b):
    try:
        resultado = a / b
    except ZeroDivisionError:
        print("No se puede dividir entre 0")
    else:
        print("Resultado:", resultado)    # solo si NO hubo error
    finally:
        print("Fin del intento")          # SIEMPRE, haya error o no

dividir(10, 2)
dividir(10, 0)
````

#### 5 · Tus propias clases de excepción

Puedes crear tus propios tipos de error heredando de `Exception` (las clases y la herencia, en el capítulo siguiente):

````python
class NotaInvalida(Exception):
    pass

def comprobar(nota):
    if not 0 <= nota <= 10:
        raise NotaInvalida(f"{nota} no está entre 0 y 10")
    return nota

try:
    comprobar(12)
except NotaInvalida as e:
    print("Error:", e)
````

#### 6 · Varios tipos a la vez, la jerarquía entera y volver a lanzar

Un mismo `except` puede atender **varios tipos** a la vez: se escriben en una tupla.

````python
def convierte(texto, divisor):
    try:
        return int(texto) / divisor
    except (ValueError, ZeroDivisionError) as e:
        print("No se puede:", type(e).__name__)
        return None

print(convierte("8", 2))
print(convierte("ocho", 2))
print(convierte("8", 0))
````

````python
def lee_nota(texto):
    try:
        return float(texto)
    except ValueError:
        print("Aviso: dato raro ->", texto)
        raise

try:
    lee_nota("siete")
except ValueError as e:
    print("Lo atiende quien llamó:", e)
````
