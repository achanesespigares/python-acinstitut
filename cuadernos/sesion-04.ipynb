#### 2 · Intercambiar dos valores

Queremos que `a` y `b` se intercambien los valores. El primer intento que se le ocurre a casi todo el mundo es este:

````python
a = 5
b = 8
a = b
b = a
print(a, b)
````

````python
a = 5
b = 8
aux = a
a = b
b = aux
print(a, b)
````

````python
a = 5
b = 8
a, b = b, a
print(a, b)
````

### Cómo funciona Python

[Abrir el capítulo en el libro](https://achanesespigares.github.io/python-acinstitut/capitulos/como-funciona-python/)

#### 1 · Pedir datos, convertirlos y mostrarlos

`input()` muestra un mensaje, espera a que el usuario escriba y pulse Intro, y devuelve lo que ha escrito. Siempre como **texto**, aunque sea un número.

````python
a = input("Primer número: ")
b = int(input("Segundo número: "))
print(a * b)
print(int(a) * b)
print(a, b, sep=" y ", end="!\n")
````

````python
print(1, 2, 3)
print(1, 2, 3, sep="-")
print("sin salto", end=" ")
print("de línea")
````
