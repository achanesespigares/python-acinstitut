#### 6 · `match`: elegir por la forma

El `match` (desde Python 3.10) compara un valor con una lista de **patrones**, de arriba abajo, y ejecuta el primero que encaja. Un patrón puede describir la forma del dato y, a la vez, **capturar** sus partes en variables.

````python
punto = (0, 3)
match punto:
    case (0, 0):
        print("el origen")
    case (0, y):
        print("en el eje vertical, a altura", y)
    case (x, 0):
        print("en el eje horizontal, en", x)
    case _:
        print("fuera de los ejes")
````

### Funciones como valores

[Abrir el capítulo en el libro](https://python.antonychanes.com/capitulos/funciones-valores/)

#### 1 · Ordenar con un criterio

`sorted` ordena de menor a mayor. Con tuplas `(nombre, nota)`, compararía primero los nombres. Para ordenar por nota, le pasamos un criterio con `key`: una función que, para cada elemento, dice **qué hay que comparar**.

````python
alumnos = [("Ana", 7), ("Biel", 9), ("Carla", 5)]
por_nota = sorted(alumnos, key=lambda a: a[1])
print(por_nota)
````

#### 2 · `map` y `filter`

Dos funciones clásicas reciben una función y una secuencia:

````python
datos = [3, 8, 11, 6]
dobles = list(map(lambda x: x * 2, datos))
pares = list(filter(lambda x: x % 2 == 0, datos))
print(dobles, pares)
````
