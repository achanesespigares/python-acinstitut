### Recorrer listas

[Abrir el capítulo en el libro](https://python.antonychanes.com/capitulos/recorrer-listas/)

#### 1 · Recorrer y sumar

`total` es un **acumulador**: empieza en 0 y en cada vuelta se le suma lo que trae `nota`. Fíjate en la tabla de traza: la columna `total` es la suma parcial hasta esa casilla.

````python
notas = [7, 4, 9, 6, 8]
total = 0
for nota in notas:
    total = total + nota
print("Suma:", total)
````

#### 2 · Recorrer por posición

`range(len(notas))` produce los números 0, 1, 2, 3 y 4: justo las posiciones de una lista de 5 casillas. El último es 4, no 5. Por eso las posiciones empiezan en 0.

````python
notas = [7, 4, 9, 6, 8]
total = 0
for i in range(len(notas)):
    total = total + notas[i]
print("Media:", total / len(notas))
````

#### 3 · El campeón provisional

El bucle empieza en la posición 1 porque la casilla 0 ya es el campeón inicial. Dentro del bucle hay un `if`: el visualizador te muestra la condición con los valores sustituidos, por ejemplo `4 > 7 → False`.

````python
notas = [7, 4, 9, 6, 8]
maximo = notas[0]
for i in range(1, len(notas)):
    if notas[i] > maximo:
        maximo = notas[i]
print("Máximo:", maximo)
````

#### 4 · Buscar un valor

Aquí el bucle es un `while`, porque no sabemos de antemano cuántas casillas hará falta mirar. En cuanto encuentra el 6, deja de buscar.

````python
notas = [7, 4, 9, 6, 8]
buscada = 6
i = 0
while i < len(notas) and notas[i] != buscada:
    i = i + 1
if i < len(notas):
    print("Está en la posición", i)
else:
    print("No está")
````

### Listas a fondo

[Abrir el capítulo en el libro](https://python.antonychanes.com/capitulos/listas-a-fondo/)

#### 1 · Índices negativos y rebanadas

Las rebanadas funcionan como en las cadenas: `lista[inicio:fin:paso]`, con el `fin` sin incluir.

````python
letras = ["a", "b", "c", "d", "e"]
print(letras[-1], letras[-2])
print(letras[1:4], letras[:2], letras[3:])
print(letras[::2], letras[::-1])
print(letras[-3:])
````

#### 2 · Alias o copia

Tres maneras de copiar una lista: `a[:]`, `list(a)` y `a.copy()`. Para comprobar si dos variables son la misma lista, usa `is`: `a is b` da `True`, `a is c` da `False`.

````python
a = [1, 2, 3]
b = a
c = a[:]
b.append(4)
c[0] = 99
print(a, b, c)
````

#### 3 · Insertar, buscar y quitar

````python
cola = ["Ana", "Biel", "Carla"]
cola.insert(1, "Dani")
cola.append("Eva")
del cola[0]
atendida = cola.pop(0)
print(cola, atendida, "Eva" in cola)
````

#### 4 · Ordenar: `sort` o `sorted`

`reverse()` le da la vuelta a una lista en su sitio; `lista[::-1]` fabrica una nueva al revés.

````python
notas = [7, 4, 9, 5]
ordenadas = sorted(notas)        # fabrica una lista nueva; notas no cambia
print(notas, ordenadas)
notas.sort(reverse=True)         # ordena notas en su sitio
print(notas)
resultado = notas.sort()
print(resultado)                 # sort no devuelve la lista: devuelve None
````
