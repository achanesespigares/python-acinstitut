#### 2 · El algoritmo de Euclides

Uno de los algoritmos más antiguos que se conocen: el máximo común divisor de dos números, descrito por Euclides hace más de dos mil años.

```python
a = 48
b = 18
while b != 0:
    r = a % b
    a = b
    b = r
print(a)
```
